```python
import pandas as pd
data = pd.read_csv(r"C:\Users\Mohana\Desktop\DE ML\iris_cluster.csv")

from sklearn.cluster import KMeans
model = KMeans(n_clusters = 10)

model.fit(data)
x = model.predict(data)

#Storing the cluster no. in the data set
data['prediction'] = x
data.to_csv(r"C:\Users\Mohana\Desktop\DE ML\KMeans_prediction.csv")
```
